```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import pysam
import annotation_plot as ap
```

# Annotation and feature tracks together (GSL & SSL based)

```python
basepath=""

vcf_path = f"{basepath}/Ychromosome/VARCALL/vcf/GSL_M_CM019820.2/GSL_M_CM019820.2.MQ30.BQ30.snps.rep_Otariidae_ZalCal.fem90.Q60_DP1000mean25_MQ60_NA10_het50_bias50_PI_50pos.vcf"
vcf = pysam.VariantFile(vcf_path)
bioinf_pos_gsl = [record.pos for record in vcf]
vcf_path = f"{basepath}/Ychromosome/VARCALL/vcf/SSL_CM019820.2/SSL_CM019820.2.MQ30.BQ30.snps._hom._repOtZC.Q60_DP300mean0.25_NA0.1_bias0.5_het0.5_PI_49pos.vcf.gz"
vcf = pysam.VariantFile(vcf_path)
bioinf_pos_ssl = [record.pos for record in vcf]

positions_csv=f"{basepath}/Pop_gen_2019/Y_haplotypes/SeaLion_AssayGenotypes_filteredOrdered_2021.6.8.positions.csv"
df_positions=pd.read_csv(positions_csv, sep='\t', usecols=[1], header=None)
ma_pos=df_positions.iloc[:, 0].astype(int).tolist()

repeats_bed = f"{basepath}/Ychromosome/lists/filtered_sites/Ychromsome_repeats_Otariidae_ZalCal.bed"
homologs_bed = f"{basepath}/Ychromosome/lists/filtered_sites/Ychromsome_homologous_regions_FEMALE.eval0.01.pident90.len20.bed"
# resulting_bed = f"{basepath}/Ychromosome/lists/filtered_sites/Ychromosome__merged.rep_Otariidae_ZalCal.fem90.subtract.bed"
coverage_bed = f"{basepath}/Ychromosome/Coverage/GSL_CM019820.2_win1/FH01.REA.MD.CM019820.2.Q20.F3844.bam.win1.mean.regions.bed.gz"
genes_csv=f"{basepath}/Pop_gen_2019/Y_annotation/Ychromsome_annotation_Rin.txt"

df = pd.read_csv(genes_csv, header=0, sep="\t")
df = df.loc[(df["gene_biotype"] != "lncRNA") & (df["gene_biotype"] != "snRNA")]

colours = {"single":"#6699CC", #blue
           "multi":"#FFCC66", #yellow
           "pseudo":"forestgreen",
           "Xtrans":"#CC3333"} #red
features_data = [
    dict(start=int(row[2]), end=int(row[3]),
         strand=(-1 if row[4]=="-" else 1),
         color=colours[row[5]])
    for _, row in df.iterrows()
]
labels = df["Name"].astype(str).tolist()
sides  = df["Placement"].astype(str).tolist()

tracks = [
    { "type":"features",
      "features": features_data,
      "labels": labels,
      "sides":  sides,
      "label": "Annotation",
      "baseline": True,
      "bar_height": 0.6,
      "head_frac": 0.10,
      "x_offsets": {2:20000, 14:13000, 58:10000}
    },
    {"type":"ticks","data": ma_pos, "label":"MassARRAY", "color":"#4a485c"},
    {"type":"ticks","data": bioinf_pos_gsl, "label":"WGS SNPs, GSL", "color":"#4a485c"},
    {"type":"ticks","data": bioinf_pos_ssl, "label":"WGS SNPs, SSL", "color":"#4a485c"},
    {"type":"intervals", "data": repeats_bed, "label":"Repeats", "color":"#4a485c"},
    {"type":"intervals", "data": homologs_bed, "label":"High-similarity\nregions", "color":"#4a485c", "label_size":11},
    {"type":"coverage_line", "data": coverage_bed, "label":"Coverage", "color":"#4a485c", "ylim":(0,100)},
    # {"type":"coverage_line", "data": coverage_bed, "label":"Coverage\n(0,10000)", "color":"#4a485c", "ylim":(0,10000)}
]


ap.plot_genomic_tracks(tracks, figsize=(17,6), dpi=1200)
```
