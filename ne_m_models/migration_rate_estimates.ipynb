##### Supplementary Text 4: Estimating the sex-biased migration via X and AUT differentiation

##### Implementation based on:
##### Sohini Ramachandran, Noah A. Rosenberg, Marcus W. Feldman, John Wakeley
##### Population differentiation and migration: Coalescence times in a two-sex island model for autosomal and X-linked loci
##### Theoretical Population Biology, 2008
##### https://doi.org/10.1016/j.tpb.2008.08.003
##### Equations (10) & (11)

```python
def get_mutation_rate(r, Fst_A, Fst_X):
    Mf = (1 / (1 - r)) * (
        (2 - r) / (6 * Fst_X) -
        (1/8) * ( (1/Fst_A) + (1/3) ) -
        (1 - r) / 6
    )
    Mm = (1 / r) * (
        -(2 - r) / (6 * Fst_X) +
        (1/4) * ( (1/Fst_A) + (1/3) ) -
        r / 6
    )
    return round(Mf, 3), round(Mm, 3), round(Mm/Mf, 3)
```

```python
def get_values(Fst_A, Fst_X, Ne_ratio):
    r = Ne_ratio / (Ne_ratio + 1)
    Mf, Mm, ratio = get_mutation_rate(r, Fst_A, Fst_X)
    print(f"SSL\nObserved values: \n\tMf: \t{Mf}, \n\tMm: \t{Mm}, \n\tMm/Mf: \t{ratio}, \n\tr: \t{r}")

    X_exp_min=(2*Fst_A*(2-r))/(3+Fst_A*(1-2*r))
    X_exp_max=(4*Fst_A*(2-r))/(3+Fst_A*(5-4*r))
    r_exp= 2 - ( 3 * (1/Fst_A-1) ) / ( 4 * (1/Fst_X-1) )
    print("Expected values:")
    print(f"\tFst_X:\t{X_exp_min} < {Fst_X} < {X_exp_max}, \n\tr: \t<{r_exp}")
```
